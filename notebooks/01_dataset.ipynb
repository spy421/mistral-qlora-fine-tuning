```python
from datasets import load_dataset 
#импортируем библиотеку для работы с датасетами
```

```python
dataset = load_dataset("databricks/databricks-dolly-15k", split="train")
#загружаем датасет
```

```python
print(dataset[0])
```

```python
print(dataset)
```

```python
print("Пустых instruction:", sum(not x["instruction"].strip() for x in dataset))
print("Пустых response:", sum(not x["response"].strip() for x in dataset))
#определяем количество пустых инструкций и ответов в датасете
```

```python
pairs = set()

for x in dataset:
    pair = (x["instruction"].strip(), x["response"].strip())
    pairs.add(pair)
print("Всего примеров:", len(dataset))
print("Уникальных примеров:", len(pairs))
#определяем количество уникальных примеров в датасете
```

```python
unique_dataset = list(set(
    (x["instruction"].strip(), x["response"].strip()) 
    for x in dataset
    ))
print("Уникальных примеров после удаления дубликатов:", len(unique_dataset))
#удаляем дубликаты из датасета
```

```python
cleaned_dataset = [
    {
        "instruction": instruction,
        "response": response
    }
    for instruction, response in unique_dataset
]
print("Количество примеров:", len(cleaned_dataset))
print(cleaned_dataset[0])
#возвращаем очищенный датасет в виде списка словарей с ключами "instruction" и "response"
```

```python
#фильтруем датасет, оставляя только те примеры, где длина инструкции не менее 10 символов, а длина ответа не менее 20 символов
filtred_dataset = [
     x 
     for x in cleaned_dataset
        if len(x["instruction"].strip()) >= 10 
        and len(x["response"].strip()) >= 20
]
print("До фильтрации:", len(cleaned_dataset))
print("После фильтрации:", len(filtred_dataset))
```

```python
for x in filtred_dataset[:3]:
    print(x)
    print("-" * 50)
```

```python
# оставляем случайные 500 примеров из отфильтрованного датасета
import random
random.seed(42)
final_dataset = random.sample(filtred_dataset, 500)

print("Финальное количество примеров:", len(final_dataset))
```

```python
print(final_dataset[0])
```

```python
# сохраняем финальный датасет в формате JSONL
import json

with open("../data/train.jsonl", "w", encoding="utf-8") as f:
    for x in final_dataset:
        f.write(json.dumps(x, ensure_ascii=False) + "\n")

print("Файл сохранён.")
```

```python
# проверяем количество строк в сохранённом файле
with open("../data/train.jsonl", "r", encoding="utf-8") as f:
    lines = f.readlines()
print("Количество строк в файле:", len(lines))
```
